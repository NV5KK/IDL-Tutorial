## 5. 演習2: 標高データの3D表示と等高線表示

ラスタ形式の標高データ（サイズ 64x64、`elevbin.dat`）をいくつかの方法で表示します。

### 5.1 データを読み込み 3D 表示する

`READ_BINARY` 関数でバイナリファイルを読み込みます。バイナリファイル自体にはデータタイプやサイズの情報がないため、あらかじめこれらを知っておく必要があります（`elevbin.dat` はバイト型・64x64）。`DATA_TYPE` キーワード省略時はデフォルトでバイト型として読み込まれます。


```idl
elev = READ_BINARY('elevbin.dat', DATA_DIMS=[64, 64])
s = SURFACE(elev)
```

マウスでオブジェクトをクリック＆ドラッグすると、サーフェスを360度回転できます。

地表画像を `TEXTURE_IMAGE` キーワードで重ねて表示することもできます。


```idl
READ_JPEG, 'elev_t.jpg', image
s = SURFACE(elev, TEXTURE_IMAGE=image)
```

### 5.2 等高線を表示する

`CONTOUR` 関数でコンタを表示します。`C_LABEL_SHOW` キーワードでラベルの表示/非表示を制御します。


```idl
c = CONTOUR(elev, C_LABEL_SHOW=0)
```

`C_VALUE` キーワードで指定した値の等高線を表示するカスタムコンタプロットも作成できます。


```idl
c = CONTOUR(elev, C_VALUE=[30, 60, 90, 120, 200], $
             C_LABEL_SHOW=[0, 1], FONT_SIZE=6)
```

> `C_LABEL_SHOW` はスカラで `1` を指定するとすべての等高線にラベルを表示します。等高線数と同数の要素を持つ配列（0/1）を指定すると、どの等高線にラベルを表示するか個別に指定できます（配列が等高線数より小さい場合は繰り返し適用）。

`RGB_TABLE` と `FILL` キーワードで等高線にカラーを付与し塗りつぶすことができます。`POSITION`（`[左下X, 左下Y, 右上X, 右上Y]`、デフォルト範囲0-1）でウィンドウ内の表示位置、`TITLE` でタイトルを指定します。


```idl
c = CONTOUR(elev, RGB_TABLE=0, C_VALUE=[0, 30, 60, 90, 120, 200], $
             /FILL, POSITION=[0.1, 0.1, 0.8, 0.8], TITLE='DISPLAY A CONTOUR')
```

> 等高線の間隔を明確に指定する場合は `C_VALUE`、任意の本数を自動間隔で描く場合は `N_LEVELS` キーワードを使用します。

最後にカラーバーを表示します。


```idl
cb = COLORBAR(TARGET=c, ORIENTATION=1, TAPER=1, $
               POSITION=[0.85, 0.1, 0.9, 0.8], TEXTPOS=1, TITLE='CONTOUR COLORBAR')
```

> IDL はグレースケール以外にも多数のカラーテーブルを搭載しています。詳細は `IDL > Graphics > Graphics Gallery > Modifying Visualizations > Fonts and Colors > Loading Default Color Tables` を参照。

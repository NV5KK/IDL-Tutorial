## 2. 演習に入る前に

IDL for VSCode のセットアップ方法と、基本的な UI・使い方について説明します。

### 2.1 セットアップ手順

1. **IDL 本体をインストールする**
   IDL 9.3 のインストーラを実行し、IDL 本体（コンパイラ・実行エンジン）をマシンにインストールします。プログラムの**コンパイルや実行、デバッグ、IDL ノートブックの実行**にはこの本体が必要です。
2. **Visual Studio Code をインストールする**
   まだ導入していない場合は https://code.visualstudio.com/ から Visual Studio Code をインストールします。
3. **IDL for VSCode 拡張機能をインストールする**
   VS Code の拡張機能ビュー（`Ctrl+Shift+X`）で「IDL for VSCode」を検索してインストールするか、VS Code のクイックオープン（`Ctrl+P`）に以下を貼り付けて実行します。
   ```
   ext install idl.idl-for-vscode
   ```
   （参考: https://marketplace.visualstudio.com/items?itemName=IDL.idl-for-vscode ）
4. **フォルダ（ワークスペース）を開く**
   VS Code の「フォルダを開く」で、演習用のデータや `.pro` ファイルを置く作業ディレクトリを開きます。拡張機能が IDL ファイルを認識し、言語サーバーが起動します。
5. **IDL Tutorials を確認する（任意）**
   拡張機能をインストールすると、サイドバーに **IDL Tutorials** の項目が追加されます。ノートブック形式のチュートリアル集で、初めて IDL に触れる場合の導入に役立ちます。

> ※ IDL Workbench は引き続き利用可能ですが、NV5 Geospatial は今後 IDL for VSCode への移行を推奨しています。本トレーニングでは IDL for VSCode を前提に説明します。

### 2.2 IDL for VSCode の UI について

VS Code 標準の UI に、IDL 用の機能が統合されています。

主な構成要素:
- **エクスプローラー**: 開いているフォルダ内のファイル（`.pro` ファイルやデータファイルなど）を表示
- **エディタ**: IDL プログラムを編集するウィンドウ。構文ハイライト、型検出に基づくオートコンプリート、コード整形に対応
- **問題（Problems）パネル**: コンパイルせずに検出された文法・型の問題を一覧表示
- **統合ターミナル**: IDL を対話的に実行できるコンソール（`IDL>` プロンプト）。VS Code のターミナルパネル内で動作します
- **実行とデバッグ（Run and Debug）ビュー**: ブレークポイントの設定や変数の確認を行うデバッグ用パネル
- **IDL ノートブック**: `.idlnb` 形式のノートブックをエディタとして開き、セル単位で IDL コードを実行しグラフィックスを埋め込み表示できる機能

### 2.3 便利な機能

**統合ターミナル（IDL コンソール）**
- ターミナル上の `IDL>` プロンプトにステートメントを入力すると、インタラクティブに変数や動作を確認できます
- カーソルキーの上矢印で入力履歴を呼び出し、そのまま実行可能
- コマンドの大文字・小文字は区別されません（IDL 内部ではすべて大文字として認識・実行）

**ホバーヘルプとオートコンプリート**
- ルーチン名や変数にマウスカーソルを乗せると、ドキュメントを含むホバーヘルプが自動的に表示されます
- 入力中に関連するルーチン・キーワード・変数の候補がオートコンプリートとして表示されます

**問題検出とコードフォーマット**
- コンパイル前にコード内の問題を検出し、修正候補を提示します
- コマンドパレット（`Ctrl+Shift+P`）からコードのフォーマットを実行できます

### 2.4 ヘルプの活用

- **ホバーヘルプ / 埋め込みドキュメント**: IDL for VSCode にはルーチンのドキュメントが同梱されており、エディタ上でホバーするだけで詳細を確認できます
- **Documentation Center**: 米国本社ホームページ上で IDL の全ドキュメントが公開されています。ブラウザの翻訳アドオンを使えば機械翻訳で概要を把握できます。
  https://www.nv5geospatialsoftware.com/docs/home.html

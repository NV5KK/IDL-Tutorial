## 8. 演習5: IDL でのプログラミング

### 8.1 IDL でのプログラミングについて

IDL アプリケーションの範囲は、コマンドラインに入力する短いプログラムから、GUI を持つ大規模プログラムまで様々です。C、C++、FORTRAN に慣れた開発者にとって IDL の構文は親しみやすいものです。

IDL が他のプログラム言語よりも優れている点:
- **配列演算**: 各データ要素への繰り返し演算が不要になり、効率的なプログラムを作成できる
- **ダイナミックデータタイプ**: コードの前後関係からデータタイプが決定されるため、変数を明示的に宣言する必要がない
- **IDL for VSCode 開発環境**: 構文ハイライト、型検出に基づくコーディング支援、問題検出、ビジュアルデバッグツール
- **対話型プログラミングモード**: コマンドラインから即座にコードをテスト可能
- **GUI ツール**: ウィジットプログラミングによるボタン・スライダ等のコントロール作成
- **組み込みルーチン**: GUI・数値解析・データ可視化用の膨大なルーチンライブラリ
- **統合型開発**: 他言語で作成された外部プログラムの呼び出し、または外部プログラムからの IDL 呼び出し
- **配布**: ソースコード、またはコンパイル済みバイナリ（SAVE ファイル）として配布可能。開発ライセンスがなくても IDL バーチャルマシンで大半のアプリケーションを実行可能
- **他言語とのリンク**: C、Java、Python などとの相互呼び出し（詳細: `IDL > IDL Bridges`）

### 8.2 IDL のステートメント

**分岐文: IF...THEN...ELSE**

条件が真なら `THEN`、偽なら `ELSE` に続くステートメントを実行します。


````idl
a = 4
IF a EQ 4 THEN PRINT, 'a is 4' ELSE PRINT, 'a is not 4'
; a is 4

a = 5
IF a EQ 4 THEN PRINT, 'a is 4' ELSE PRINT, 'a is not 4'
; a is not 4
````

**ループ文: FOR**

初期値・最大値・インクリメント値を設定し、`DO` に続くステートメントをループ回数分実行します。


````idl
FOR count = 1, 5 DO PRINT, 'count =', count
; count =       1
; count =       2
; count =       3
; count =       4
; count =       5
````

**複合ステートメント**

分岐文・ループ文の実行部分が複数行にわたる場合、`BEGIN...END`（`ENDIF`, `ENDELSE`, `ENDFOR` など）が必要です。ただしコマンドラインからは入力できず、プログラム内でのみ使用可能です。

> その他の分岐文・ループ文は `IDL > Routines (by topic) > Statements` を参照。

**条件文で使用する演算子**

| 演算子 | 説明 | 例 |
|---|---|---|
| EQ | ～に等しい | `PRINT, 5 EQ 2, 2.0 EQ 2` → `0 1` |
| NE | ～に等しくない | |
| LE | ～以下 | |
| LT | ～未満 | |
| GE | ～以上 | |
| GT | ～より大きい | |
| AND | ビット積 | `PRINT, 5 AND 6` → `4` |
| OR | ビット和 | |
| && | 論理積 | |
| \|\| | 論理和 | |

> 詳細は `IDL > IDL Concepts > IDL Programming > Operators` を参照。

### 8.3 IDL のプログラミングタイプ

IDL には **メインプログラム**、**プロシージャ**、**関数** の3つのプログラミングタイプがあります。

**メインプログラム**

別ファイルを作成せず、いくつかのコマンドを実行する場合に使用します。プロシージャ (`PRO`) や関数 (`FUNCTION`) のように先頭で明示的に指定されませんが、`END` ステートメントは必要です。他のルーチンから呼び出したり引数を受け取ることはできません。`.RUN` コマンドでコマンドラインから実行するか、ファイルに保存して実行します。


````idl
; .RUN を使用してコマンドラインから実行する例
; IDL> .RUN
; - x = FINDGEN(100)
; - y = SIN(x)/10
; - p1 = PLOT(x, y)
; - END
````

**プロシージャと関数**（名前付きプログラム）

プロシージャと関数は共にモジュラープログラムで、個別実行や他プログラムからの呼び出しが可能です。1つのファイルに複数のプロシージャ・関数を含めることもできます。

- **プロシージャ**: `PRO <プロシージャ名>` で定義。戻り値なし。
- **関数**: `FUNCTION <関数名>` で定義。`RETURN` ステートメントで値を返す。デフォルトで新しい変数を作成するため、戻り値へのアクセスが必要な場合は関数を使用する。


````idl
TEST          ; プロシージャの呼び出し例
data = TEST()   ; 関数の呼び出し例
````

> 関数は実行アイコンなどから直接実行することができません。コンソール上で名前を指定して呼び出す必要があります。
> サンプルプログラムは IDL インストールディレクトリ配下の `/examples/demo/demosrc` にあるデモプログラムを参照できます（変更・保存はしないでください）。

### 8.4 引数とキーワード（プロシージャ・関数）

プロシージャ・関数名の後にカンマ区切りでパラメータを並べます。キーワードは省略可能なパラメータで、プログラム内部で制御が必要です。


````idl
TEST1, 1, 2, Key=3
; prm1 =        1
; prm2 =        2
; prm3 =        3

ret = TEST2(1, 2, Key=3)
; prm1 =        1
; prm3 =        3
PRINT, ret
;        3
````

> `KEYWORD_SET` 関数はキーワードがセットされているかチェックします（セット時 1、未セット時 0 を返す）。詳細は `IDL > Routines (alphabetical) > Routines: K > KEYWORD_SET` を参照。

### 8.5 IDL のプログラムとコンパイル

**プログラムファイル**

IDL のプログラムファイルは拡張子 `.pro`。ファイル名はファイル内のプロシージャ／関数名と一致させます。1ファイルに複数のプロシージャ・関数を記述可能ですが、ファイル名となるメインのルーチンは**ファイルの最後**に記述する必要があります。

**自動コンパイル**

ルーチンがすでに IDL のメモリにある場合はそのまま実行されます。メモリにない場合、検索パス内で該当ファイル（`ファイル名.pro`）を探し自動的にコンパイルします。ファイルの先頭から順にコンパイルし、ファイル名と同名のルーチンが見つかった時点で停止するため、メインルーチンは最後に配置します。

> 検索パスはユーザが追加・編集可能です（`IDL > IDL Concepts > Setting Preferences > Directory and Search Path Preferences`、`IDL_PATH`）。
> ユーザ作成ルーチンが IDL 固有のルーチンと同名の場合、自動コンパイルは IDL 固有のルーチンを優先します。

**実践: CONTOUR_COLORBAR プログラムの作成**

標高データを使用し、コンタカラーの標高データとカラーバーを表示するプログラムを作成します。VS Code で新しいファイルを作成し、拡張子 `.pro` で保存できるエディタを開き、以下を入力してください。


````idl
PRO CONTOUR_COLORBAR
    ; 変数 file へファイルを設定します
    file = DIALOG_PICKFILE(FILTER='elevbin.dat')
    ; dem 変数へ elevbin.dat をインポートします
    dem = READ_BINARY(file, DATA_DIMS=[64, 64])
    ; CONTOUR プロシージャを利用し、データを表示します
    c1 = CONTOUR(dem, RGB_TABLE=30, /FILL, PLANAR=0, $
                 N_LEVELS=5, TITLE='Contour and Colorbar')
    ; 表示したデータの横にカラーバーを追加します
    cbar = COLORBAR(TARGET=c1, ORIENTATION=1, TAPER=1, $
                     POSITION=[0.90, 0.2, 0.95, 0.75])
END
````

> IDL コードでセミコロン (`;`) はコメント行の開始を表します（IDL 自身はこの行を実行しません）。

**コンパイルと実行**

1. `Ctrl+S`（Mac: `Cmd+S`）で `contour_colorbar.pro` として保存します
2. 統合ターミナルの `IDL>` プロンプトから `.compile` コマンドでコンパイルします（IDL の検索パス上にあれば自動的にコンパイルされます）
3. プロシージャ名を指定して実行します

```
IDL> .compile contour_colorbar
IDL> CONTOUR_COLORBAR
```

> エラーが発生した場合は問題（Problems）パネルやコンパイル時のメッセージを確認し、コードに誤字がないか確認してください。ソースコードを変更した場合は再度 `.compile` を実行してください。

### 8.6 デバッグとブレークポイント

IDL for VSCode には、コードのエラーを検索・修正するための VS Code 標準のデバッグ機能が統合されています。

**デバッグの基本操作**

VS Code の「実行とデバッグ（Run and Debug）」ビューを使用し、他の言語と同様の操作感でプログラムをデバッグできます。

| 操作 | 内容 |
|---|---|
| デバッグ開始 | 「実行とデバッグ」ビューまたはコマンドパレットからデバッグを開始し、プログラムを実行 |
| 継続 | ブレークポイントで一時停止したプログラムを再開 |
| 停止 | 実行中のプログラムを停止 |
| ステップイン | 関数・プロシージャの処理内にステップイン |
| ステップオーバー | 関数・プロシージャを1行とみなし、その処理内には入らず次の行へ |
| ステップアウト | 実行中の関数・プロシージャを抜ける（リターン） |

**ブレークポイント**

デバッグ作業でプログラム実行中に意図的に一時停止させるポイントです。設定方法:
- エディタのコード行番号の左側の余白をクリックしてオン/オフを切り替え（VS Code 標準のブレークポイント操作と同様）

設定すると、エディタ左側の余白に赤い点が表示されます。

**変数の値を表示する**

プログラムが一時停止したら、「実行とデバッグ」ビューの変数（Variables）パネル、デバッグコンソールでの `HELP` や `PRINT` の実行、またはエディタで変数上にマウスポインタを置くことで、現在の実行範囲での変数の値を調べられます。

**一時停止中に変数を可視化する**

ブレークポイントで一時停止中に、デバッグコンソールやターミナルから `PLOT` や `IMAGE` 関数を実行することで変数を可視化でき、デバッグ効率をさらに高めることができます。

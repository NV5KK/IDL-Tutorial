## 1. IDL について

### 1.1 IDL の概要

**IDL: Interactive Data Language**

- データ解析や可視化など多種多様なルーチンを搭載
- クロスプラットフォーム対応（Windows / Mac / Linux）
- 配列指向型言語で大容量データのハンドリング効率化
- GUI アプリケーションの開発機能と無償実行環境（IDL Virtual Machine）の提供

### 1.2 開発環境の現状: IDL Workbench から IDL for Visual Studio Code へ

長らく IDL の統合開発環境（IDE）としては、Eclipse ベースの **IDL Workbench**（IDLDE）が使われてきました。しかし現在は、無償の Visual Studio Code 拡張機能である **IDL for Visual Studio Code**（以下 IDL for VSCode）への移行が公式に進められています。

> 2024年7月、NV5 Geospatial は IDL for VSCode の正式リリースを発表しました。開発の目的は「IDL プログラマーの生産性向上」と「利用の敷居を下げること」であり、今後 IDL Workbench に代わる標準の開発環境として位置付けられています。
> 参考: [Back to the Future: IDL for Visual Studio Code](https://www.nv5geospatialsoftware.com/Support/Maintenance-Detail/back-to-the-future-idl174-for-visual-studio-code)

**IDL for VSCode の主な特長**
- プログラムの**コンパイルや実行、デバッグ、IDL ノートブックの実行**には、同じマシンに IDL 本体（IDL 8.8.0 以降。IDL 9.3 ももちろん対応）のインストールが必要です。拡張機能単体はエディタ機能（構文ハイライトやオートコンプリートなどのコード編集支援）を提供するものであり、実際にコードを動かすには IDL 本体が不可欠です
- データ型検出に基づく構文ハイライト・オートコンプリート
- 最新の IDL / ENVI ドキュメントが統合されたホバーヘルプ
- コンパイルせずに 100 を超える問題を検出する静的解析（問題検出）機能
- コードの体裁を自動的に整えるコードフォーマッタ
- ブレークポイントに対応した統合デバッガー、および統合ターミナルでのコマンドライン IDL 実行
- グラフィックスの埋め込みや PDF 保存に対応した **IDL ノートブック** をネイティブサポート
- 初学者向けの **IDL Tutorials**（サイドバーから開けるノートブック形式のチュートリアル集）を同梱

> 本トレーニングでは、この **IDL for VSCode** を使用します。導入手順は次章「2. 演習に入る前に」で説明します。

**Python との連携強化**（IDL 9.2 以降、継続的に拡充）

IDL と Python の連携は IDL 9.2 以降、バージョンを重ねるごとに強化が続けられています。

IDL 9.2 での強化点:
- Python 3.13 を含む埋め込み Python 環境を標準搭載
- `PyUtils` クラスで pip によるパッケージ管理が可能（例: `PyUtils.PipInstall('beautifulsoup4')`）
- `PyUtils.Load` により外部 Python 環境の切り替えが容易に
- `Python::GetAttr` により属性を文字列で取得可能（`/WRAP` オプションで Python オブジェクトのまま取得）

IDL 9.3 での強化点（[What's New in IDL](https://www.nv5geospatialsoftware.com/docs/WhatsNew.html) より）:
- 埋め込み Python 環境が **Python 3.14** に更新
- Python ブリッジ実行時のエラー報告が改善され、Python コード内でエラーが発生した際に、最上位レベルのメッセージだけでなくスタック情報を含む**完全なトレースバック**が得られるようになった

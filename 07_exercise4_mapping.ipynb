## 7. 演習4: マッピング

IDL のマッピング機能により、様々な投影図法でデータをプロットできます。この章では地図の表示、地図画像や大陸・ベクタデータの重畳表示を行います。

### 7.1 地図をプロットする

`MAP` 関数の第一引数に投影法を指定します。


```idl
m = MAP('Mollweide')
```

> 投影法の詳細は `IDL > Routines (alphabetical) > Routines: M > MAP`、マッピングルーチン全般は `IDL > Routines (by topic) > Mapping` を参照。

### 7.2 地図に画像を重ねて表示する

世界の大陸イメージ画像（`Day.jpg`）を読み込みます。


```idl
READ_JPEG, 'Day.jpg', day
dayimg = IMAGE(day)
```

`IMAGE` 関数の `MAP_PROJECTION` キーワードで画像を地図に重ねます。`GRID_UNITS` でグリッド単位（ここでは緯度経度）、`IMAGE_LOCATION` で画像左下の地図座標、`IMAGE_DIMENSIONS` で描画範囲を指定します。


```idl
map = IMAGE(day, LIMIT=[-90, -180, 90, 180], GRID_UNITS=2, $
             IMAGE_LOCATION=[-180, -90], IMAGE_DIMENSIONS=[360, 180], $
             TRANSPARENCY=50, MAP_PROJECTION='Mollweide')
```

`MAP_PROJECTION` プロパティを変更するだけで投影法を切り替えられます。


```idl
map.MAP_PROJECTION = 'Equirectangular'
cont = MAPCONTINENTS(/COUNTRIES)
```

### 7.3 地図にベクタデータを重ねて表示する

風速・風向のベクタデータ（`globalwinds.dat`、事前に `SAVE` プロシージャでパッケージ化済み）を `RESTORE` で読み込みます。


```idl
RESTORE, 'globalwinds.dat'
```

> `SAVE`/`RESTORE` の詳細は `IDL > Routines (alphabetical) > Routines: S > SAVE` を参照。

`VECTOR` 関数で風のベクタデータを地図に重畳します（引数: u=水平成分, v=垂直成分, x=経度, y=緯度）。


```idl
vect = VECTOR(u, v, x, y, /OVERPLOT, RGB_TABLE=39, $
               AUTO_COLOR=1, HEAD_ANGLE=15, HEAD_SIZE=0.8)

map.limit = [30, 127, 48, 147]   ; 日本付近のみ表示
```

> `AUTO_COLOR=1` によりベクトルの大きさに応じたカラーが付与されます。詳細は `IDL > Routines (alphabetical) > Routines: V > VECTOR` を参照。

最後に `STREAMLINE` 関数でベクタデータを流線形に重ねます（先に `DELETE` メソッドでベクタ表示を削除）。


```idl
vect.delete
stream = STREAMLINE(u, v, x, y, /OVERPLOT, $
                     TRANSPARENCY=20, RGB_TABLE=33, AUTO_COLOR=1, THICK=5)
```

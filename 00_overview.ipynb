# IDL 9.3 入門トレーニング

NV5 Geospatial 株式会社

はじめて IDL に触れる方が、基本的な機能や文法事項を一通り整理できることを目的とした入門教材です。
IDL の基本概念から、画像処理・3D表示・信号処理・マッピング・プログラミングまで、5つの演習を通じて学習します。
開発環境には、現在 IDL の標準的な開発環境として利用が進められている **IDL for Visual Studio Code** を使用します。

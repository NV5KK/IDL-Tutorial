## 6. 演習3: 信号処理

この章では IDL でのデジタル信号のノイズ除去・フィルタリングについて学習します。ノイズを追加したデータセットを作成し、`SMOOTH` によるノイズ除去と、`FFT` を用いた周波数領域フィルタ処理（ローパス/ハイパス）を行います。分かりやすさのため1次元信号のみを扱います。

IDL には基本的なデジタルフィルタルーチン（`SMOOTH` など）から、ウェーブレットフィルタリング（`WV_DENOISE`）のような複雑なルーチンまで多数用意されています。

### 6.1 データセットの作成

時間と共に増加する周波数を持つ正弦波関数を作成します。


```idl
data = SIN((FINDGEN(200) / 35) ^ 2.5)
p = PLOT(data)
```

> `SIN` 関数は引数（ラジアン単位）の正弦値を返します。`FINDGEN` はそれぞれの添え字の値を要素に持つ浮動小数点配列を返します。数学ルーチンの詳細は `IDL > Routines (by topic) > Mathematics` を参照。

一様分布のランダムノイズを追加します。


```idl
noisy = data + ((RANDOMU(SEED, 200) - 0.5) / 2)
p = PLOT(noisy)
```

オリジナルとノイズ付きデータを重ねて表示します（`OVERPLOT` キーワード）。


```idl
p = PLOT(data, XTITLE='Time', YTITLE='Amplitude', THICK=3)
p = PLOT(noisy, COLOR='red', /OVERPLOT)
```

> `PLOT` 関数はデータ可視化の基本ルーチンです。詳細は `IDL > Routines (alphabetical) > Routines: P > PLOT` を参照。

### 6.2 SMOOTH 関数での信号処理

`SMOOTH` 関数は、指定幅の移動（boxcar）平均によって平滑化された配列を返します。


```idl
smoothed = SMOOTH(noisy, 5)
s = PLOT(smoothed, TITLE='Smoothed Data')
```

SMOOTH がノイズを除去すると、周波数が増加するにつれ振幅が減少していくことが確認できます。

### 6.3 周波数領域フィルタ処理

ノイズを含むデータにローパスフィルタ（低域通過フィルタ）を適用することで、高周波成分（ノイズ）を軽減します。


```idl
y = [FINDGEN(100), FINDGEN(100) - 100]
y[101:199] = REVERSE(y[0:98])
filter = 1.0 / (1 + (y / 40) ^ 10)
f = PLOT(filter)
```

フィルタの周波数特性にデータの FFT 成分を掛け算し、逆フーリエ変換で空間領域に戻すことでフィルタリングします。


```idl
lowpass = FFT(FFT(noisy, 1) * filter, -1)
i = PLOT(lowpass)
```

同じフィルタ機能をハイパスフィルタ（高周波・ノイズ要素のみ通過）として使うこともできます。


```idl
highpass = FFT(FFT(noisy, 1) * (1.0 - filter), -1)
h = PLOT(highpass, YRANGE=[-0.4, 0.3])
```
